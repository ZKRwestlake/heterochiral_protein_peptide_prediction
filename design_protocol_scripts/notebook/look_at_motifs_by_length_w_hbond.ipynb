```python
file = "cluster_results.dat"

df = pd.read_csv(file, sep="\s+")
# df['description'] = df['name'] + "_og"
print(len(df))
```

```python
df['super_ddg'] = -2 * df['best_hbond_to_crit'] + df['best_ddg']
def get_good_motifs(subdf, size):
#     subdf = df

    subdf = subdf[subdf['best_buried_crit_unsat'] <= 100]
    df1 = subdf.nlargest(size, 'best_hbond_to_crit')
    df2 = subdf.nsmallest(size, 'best_ddg')
    df3 = subdf.nsmallest(size, 'super_ddg')
#     display(df1.index.values)
#     display(df2.index.values)
#     display(len(df))
#     df2 = df2[['description']]
    arr = np.zeros((len(df)), dtype=np.bool)
    arr[df1.index.values] = True
    arr[df2.index.values] = True
    arr[df3.index.values] = True
    asdf = subdf.loc[arr[subdf.index.values]]

    return asdf
```

```python
!mkdir motifs_by_size_H

to_use = 0
for i in range(5, 12):
    
    subdf = df[(df['best_length'] == i) & (df['sec_type'] == 'H')]
    subdf = get_good_motifs(subdf, 40)
    
    print("%i -- %5i"%(i, len(subdf)))
    
    if (len(subdf) == 0):
        continue
        
    to_use += len(subdf)

    
    with open("motifs_by_size_H/%i.list"%(i), "w") as f:
        for idx, row in subdf.iterrows():
            f.write("%s\n"%(row['path']))

print(to_use)
```
